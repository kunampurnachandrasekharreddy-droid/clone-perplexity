```python
#using arthematic operators
x=int(input("enter any intrger:"))
y=int(input("enter any intrger:"))
print(x+y,"is addition")
print(x-y,"is subtraction")
print(x*y,"is multiplication")
print(x/y,"is division")
print(x%y,"is modulus")
print(x**y,"is power")
print(x//y,"is floor division")

```

```python
#comparision operator
x=int(input("enter any intrger:"))
y=int(input("enter any intrger:"))
print(x==y)
print(x!=y)
print(x>y)
print(x<y)
print(x>=y)
print(x<=y)
```

```python
#logical oiperators
x=int(input("enter any intrger:"))
y=int(input("enter any intrger:"))
print(x>y and x<y)
print(x>y or x<y)
print(not(x>y))
```

```python
#bitwise operators
x=int(input("enter any intrger:"))
y=int(input("enter any intrger:"))
print(x>y & x<y)
print(x>y | x<y)


```
