```python
#check given year is leap year or not
a=int(input("enter any year:"))
if(a%4==0 and a%100!=0)or(a%400==0):
  print(a,"is a leap year")
else:
  print(a,"not a leap year")
```

```python
#example for the match case
a=int(input("enter:"))
match a:
  case 1:
    print("monday")
  case 2:
    print("tuesday")
  case 3:
    print("wedensday")
  case 4:
    print("thursday")
  case 5:
    print("friday")
  case 6:
    print("saturday")
  case 7:
    print("sunday")
  case _:
    print("invalid")
```

```python
#to find sum of the digits in a number
a=int(input("enter a:"))
sum=0
while(a!=0):
  b=a%10
  sum=sum+b
  a=a//10
print(sum)
```

```python

#program that checks the prine number or not
a=int(input("enter a numb:"))
count=0
for i in range(2,a//2+1):
  if a%i==0:
    count+=1
    break
if(count==0 ):
  print("prime number")
else:
  print("not a prime number")
```

```python
#program that prints prime numbers bwn given two num
a=int(input("enter first numb:"))
b=int(input("enter last numb:"))
for i in range(a,b+1):
    for j in range(2,i):
        if(i%j==0):
            break
    else:
        print(i,end=(" "))
```
